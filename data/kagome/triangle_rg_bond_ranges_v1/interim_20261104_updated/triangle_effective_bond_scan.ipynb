# What was calculated, and what the results mean

We sample **all bonds** from the distributions below. Within each size we reuse
one base disorder realization and multiply every bond by the stated factor.
Internal triangle bonds and boundary bonds follow the same distribution.

| Factor / mean K | Distribution of each positive antiferromagnetic K |
| ---: | :--- |
| 1.0 | Uniform(0.5, 1.5) |
| 0.5 | Uniform(0.25, 0.75) |
| 0.2 | Uniform(0.10, 0.30) |
| 0.1 | Uniform(0.05, 0.15) |

The Hamiltonian is $H=\sum K_{ij}Z_iZ_j-h\sum Z_i-\Gamma\sum X_i$.
The code stores $J_{ij}=-K_{ij}$. All results here use disorder60000,
periodic kagome geometry, triangle orientation0 and superblock origin(0,0).
This is the **all-bond** study; the intertriangle-only study has its own notebook.

**The plotted quantity:** draw two independent physical spin configurations
$z^{(a)},z^{(b)}$ from the reconstructed RG ground-state probability, then compute

$$q=\frac1{48}\sum_{i=1}^{48}z_i^{(a)}z_i^{(b)}.$$

At q=1 all spins agree; at q=-1 all oppose; q near zero means roughly half
agree. The vertical axis is **probability mass in a discrete q bin**, not a
continuous probability density. L4 has 48 physical spins, so q spacing is 1/24.
Every curve uses 8192 independent replica pairs. Lines join bins as a visual aid.

**Two distinct comparisons answer our research question:** changing fields at
fixed RG retention shows whether P(q) moves; changing retention at fixed fields
shows how sensitive that result is to the approximation. The main L4 plots
contain RG curves. Exact curves appear only in the separate L2 checks.

# Lowering all bond strengths: L4 overlap distributions

**Interim results — the finite scan is still running.**

We lower both endpoints of every antiferromagnetic bond distribution and scan
the transverse field $\Gamma$ and longitudinal field $h$. The goal is a field
window where physical $P(q)$ shifts while RG remains stable under retention
changes, before finite-size scaling of $q_{\rm EA}$ and correlation length.

The model is $H=\sum_{ij}K_{ij}Z_iZ_j-h\sum_i Z_i-\Gamma\sum_i X_i$.
All bonds use $K=sK^{(0)}$, $K^{(0)}\sim U(0.5,1.5)$, with
$s=1,0.5,0.2,0.1$. Within each size, the same disorder60000 base draws are
used throughout. Internal and boundary bonds receive the same factor.

Each L4 system has 48 physical spins. Rows are $\Gamma=0.05,0.10,0.25,0.50$;
columns are $h=0,0.025,0.075,0.15$. Curves compare $(m,\chi)=(4,8),(6,8),(6,16)$,
using two Lanczos starts and 8192 independent physical replica pairs per curve.
Error bars are pointwise ±1 sampling SE; they exclude truncation uncertainty.
At zero field the solvers restrict to even spin-flip parity, avoiding arbitrary
mixtures when the tunnelling splitting is below numerical resolution.

**Interpretation:** a candidate requires TV(m), TV(chi)<0.05, chi energy change/spin/mean K<0.001, chi q2 change<0.01, matching exact L2 m6 screens, and no solver, cutoff or sampling alarms. Normalizing by mean K prevents a smaller energy unit from creating an artificial pass. Histogram TV contains sampling noise. Passing is a retention sensitivity screen, not a bound on physical error. This one-disorder snapshot does not establish a transition or validate finite-size scaling. The field background of q_EA and physical connected correlations must be treated explicitly.
