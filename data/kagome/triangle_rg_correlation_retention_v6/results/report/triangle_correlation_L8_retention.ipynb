## Interpretation

Compare m at fixed chi and chi at fixed m. Inspect each disorder as well as the ratio of averaged spectra. A systematic plateau shift with retention indicates truncation sensitivity. Agreement across this small ladder is evidence of stability within these settings, not a physical convergence proof. The former 0.01 criterion for R is descriptive here; two disorders do not qualify production curves. New settings must pass exact reduced-space L4 checks before L8; solver failures are preserved and excluded from physical comparisons.
