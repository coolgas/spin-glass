# Exact L2 versus triangle RG: varying Gamma at fixed lambda

The primary comparison is now the **same 12-spin physical model**, solved in
the full spin space and after retaining m states per triangle. Each lambda
section below compares exact and RG P(q) at Gamma=0.25, 0.5, 1 and 2, with
longitudinal field h=0.15. All curves at a given parameter pair share the
**identical saved bonds**. This is one disorder realization (disorder60000).

We first sample positive antiferromagnetic bonds independently from
$K^{(0)}_{ij}\sim U(0.5,1.5)$. Choose disjoint orientation0 triangles, call
their internal bonds A and their connecting bonds B, and use

$$H(\lambda)=\sum_{(ij)\in A}K^{(0)}_{ij}Z_iZ_j
+\lambda\sum_{(ij)\in B}K^{(0)}_{ij}Z_iZ_j
-0.15\sum_i Z_i-\Gamma\sum_i X_i.$$

| lambda | K inside triangles | K connecting triangles | Physical meaning |
| ---: | :--- | :--- | :--- |
| 0 | U(0.5, 1.5) | zero | independent frustrated triangles |
| 0.1 | U(0.5, 1.5) | U(0.05, 0.15) | weakly connected triangles |
| 0.3 | U(0.5, 1.5) | U(0.15, 0.45) | intermediate connection |
| 1 | U(0.5, 1.5) | U(0.5, 1.5) | original kagome bond law |

No bonds are redrawn as Gamma or lambda changes. L2 has 24 bonds: 12 A and
12 B, with two of each touching every spin. Lambda is dimensionless; fields
are in units of the mean internal K, which is 1. Stored couplings use J=-K.
Changing lambda changes the Hamiltonian; increasing m improves its approximation.

| Calculation | Physical spins | Triangle sites | Total solved dimension |
| :--- | ---: | ---: | :--- |
| Full L2 | 12 | — | 2^12 = 4096 |
| RG m=4 | 12 | 4 | 4^4 = 256 |
| RG m=6 | 12 | 4 | 6^4 = 1296 |
| Full triangle space m=8 | 12 | 4 | 8^4 = 4096 |

This L2 test has one projection step. The later superblock retention chi is
not used. At m=8 no triangle state is discarded, so exact recovery checks
the projection and reconstruction.

The measured quantity is the **physical** overlap between two independent
replicas drawn from the ground-state Born distribution:

$$q=\frac1{12}\sum_i z_i^{(a)}z_i^{(b)},\qquad
P(q)=\sum_{z,z'}|\psi(z)|^2|\psi(z')|^2
\mathbf{1}\!\left[q=\frac1{12}\sum_i z_i z'_i\right].$$

There are 13 bins, with spacing 1/6. The vertical axis is probability mass
per bin. RG wavefunctions are lifted back to all 12 physical spins before
measuring q. Every probability is evaluated by exact summation over the saved
states, so these L2 curves have **no replica sampling noise**. Lines join
discrete bins. The full spin reference is a numerical ground-state eigensolve.
At h=0.15, mean q includes local polarization:
$\langle q\rangle=12^{-1}\sum_i\langle Z_i\rangle^2$.
Neither this mean nor a finite-size peak alone establishes glass order.

## Does smaller lambda help convergence?

At fixed Gamma and m, weaker connecting bonds usually reduce the error here. At lambda=0 the ground state is a product of triangle ground states, already contained in each retained space, and RG is exact up to numerical precision. Increasing m from 4 to 6 markedly improves the connected cases. The two columns share color limits within each row; numbers show small errors that the common color scale can hide. Energy errors are in units of mean internal K=1.

This gives a useful controlled starting point: connecting bonds mix local triangle states, so weakening them suppresses that mixing when the relevant excitation denominators stay finite. Small bare lambda alone is not a guarantee near small denominators, nor does it make an iterative solver necessarily faster. The numerical question here is convergence of the retained approximation.

**Why it matters:** lambda allows a direct test from independent triangles to the original lattice. It is a promising route to a validated weak-coupling effective theory. To claim accuracy for the original kagome model we must still demonstrate convergence at lambda=1, and then across larger sizes. A small-lambda physical result describes the model with those weaker bonds.

# Weak-intertriangle-coupling screen for spatial RG

The eventual goal is finite-size scaling of $q_{\rm EA}$ and the physical
correlation length in a validated RG regime. This first screen asks whether
physical $P(q)$ still moves with field where retained-space sensitivity is small.

The 4 × 4 layout follows the selected comparison in
`exact_guide_overlap_distributions.ipynb`. Here rows are Gamma=.25,.5,1,2;
columns are lambda=0,.1,.3,1; h=.15. L4 means 48 physical kagome spins.
Original bonds are fixed (disorder60000); only bonds between orientation0
triangles are multiplied by lambda. Internal bonds and fields are unchanged.
The same physical interpolation is used by every retained schedule.



L4 curves compare (m,chi)=(4,8),(6,8),(6,16), with origin(0,0).
Each effective Hamiltonian is solved by Lanczos from two independent starts.
Each curve uses 8192 independent physical replica pairs in two measurement
batches. Bars are pointwise ±1 sampling SE and exclude truncation error.
The black curve at lambda=0 is an exact full-spin product-triangle reference.
At lambda>0 no exact full-space N48 solution or matched GFMC curve is supplied.
Agreement between retained spaces is a sensitivity screen, not an accuracy bound.



The companion uses exact full-space L2 references at identical model parameters
and checks the first projection at m=4,6,8. L2 disorder60000 is a different bond
realization from L4 disorder60000 under the size-dependent disorder generator.
L2 agreement does not certify the L4 superblock truncation. All m=8 projections
must recover their exact reference. Independent physical action tests verify
the bond weighting and projection.



Field shifts are visible directly in the overlay; `assessment.json` stores
paired field changes, their sampling SE, per-panel TV sensitivity, energy and
q2 changes, and exact L2 errors. Its candidate flag is provisional: unresolved
cutoffs or failed screens exclude a panel, and nonpassing points are retained.
The decoupled endpoint has field-driven local changes but no collective
intertriangle transition; motion of P(q) alone does not establish glass order.

Next validate candidate intervals with higher retention, shifted superblocks,
held-out disorders and matched full-spin L4 benchmarks. Refine Gamma/lambda
near the surviving interval before a disorder-averaged size study. At h=.15,
define q_EA with its longitudinal-field background explicitly; a finite-size
raw mean overlap or peak is not automatically a thermodynamic glass order
parameter. Define correlation length from physical connected correlations,
using one fixed Fourier/sublattice convention. Scaling uses physical L and
identical blocking rules, with separate disorder and truncation uncertainties.
