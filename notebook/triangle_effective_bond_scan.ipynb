# What was calculated, and what the results mean

We sample **all bonds** from the distributions below. Within each size we reuse
one base disorder realization and multiply every bond by the stated factor.
Internal triangle bonds and boundary bonds follow the same distribution.

| Factor / mean K | Distribution of each positive antiferromagnetic K |
| ---: | :--- |
| 1.0 | Uniform(0.5, 1.5) |
| 0.5 | Uniform(0.25, 0.75) |
| 0.2 | Uniform(0.10, 0.30) |
| 0.1 | Uniform(0.05, 0.15) |

The Hamiltonian is $H=\sum K_{ij}Z_iZ_j-h\sum Z_i-\Gamma\sum X_i$.
The code stores $J_{ij}=-K_{ij}$. All results here use disorder60000,
periodic kagome geometry, triangle orientation0 and superblock origin(0,0).
This is the **all-bond** study; the intertriangle-only study has its own notebook.

**The plotted quantity:** draw two independent physical spin configurations
$z^{(a)},z^{(b)}$ from the reconstructed RG ground-state probability, then compute

$$q=\frac1{48}\sum_{i=1}^{48}z_i^{(a)}z_i^{(b)}.$$

At q=1 all spins agree; at q=-1 all oppose; q near zero means roughly half
agree. The vertical axis is **probability mass in a discrete q bin**, not a
continuous probability density. L4 has 48 physical spins, so q spacing is 1/24.
Every curve uses 8192 independent replica pairs. Lines join bins as a visual aid.

**Error-bar convention:** production plots show only standard errors across
independent disorders. This notebook contains one disorder, so its curves
have no error bars. No disorder average or disorder uncertainty is estimated
here. Sampling diagnostics remain in the raw records; truncation sensitivity
is assessed separately. Exact L2 probabilities are computed deterministically.

**Two distinct comparisons answer our research question:** changing fields at
fixed RG retention shows whether P(q) moves; changing retention at fixed fields
shows how sensitive that result is to the approximation. The main L4 plots
contain RG curves. Exact curves appear only in the separate L2 checks.

# Lowering both endpoints of every bond distribution

The goal is to locate a candidate field region where RG is stable and physical
$P(q)$ changes, before finite-size scaling of $q_{\rm EA}$ and correlation length.
All bonds use $K=sK^{(0)}$, with $K^{(0)}\sim U(0.5,1.5)$ and
$s=1,0.5,0.2,0.1$. Internal and boundary bonds receive the same factor.
For each size, all contexts share one archived base disorder realization.

Rows: $\Gamma=0.05,0.10,0.25,0.50$. Columns: $h=0,0.025,0.075,0.15$.
L4 contains 48 physical spins. Curves compare $(m,\chi)=(4,8),(6,8),(6,16)$;
8192 independent physical replica pairs estimate each histogram. No error
bars are plotted because this study has only one disorder. Production error
bars are reserved for standard errors across independent disorders. Sampling
diagnostics are preserved in the raw records. Exact L2 probabilities are deterministic.
The L2 companion compares the first projection with exact full-spin results;
its disorder differs from L4 because the seed includes size. No full-spin L4
reference is supplied. Retention agreement is a sensitivity screen.

Lowering $K$ at fixed fields changes $\Gamma/K,h/K$. Rescaling all three
together preserves the ground state and truncation difficulty. Energy screens
below divide by the mean bond scale to avoid changing the tolerance merely
by changing the energy unit. At $h=0$, symmetry enforces zero mean overlap for
a nondegenerate exact finite-system ground state, while $P(q)$ and $q^2$ can
still change. Cutoff degeneracies, solver-start disagreement and sampling
alarms are saved and exclude provisional candidates.

## Zero-field convention and direct field overlays

At $h=0$, every exact and retained eigensolve explicitly restricts global
spin-flip parity to the even sector. This avoids arbitrary asymmetric mixtures
when the finite-system tunnelling splitting lies below numerical resolution.
The archived exact gap is the unrestricted gap, whereas retained gaps at zero
field refer to the even sector. They should not be compared as the same gap.
The unrestricted pilot is preserved separately. At finite longitudinal field,
solves use the full retained space. Near-degenerate cutoff subspaces and solver
start disagreements are retained as alarms.

The overlays use the same saved $P(q)$ measurements as the preceding panels.
They show motion directly; they do not include truncation uncertainty. Consult
the retention tables before using a shift to select a candidate window.
